```python
import gc
from minicheck.minicheck import MiniCheck
import os

import torch
os.environ["CUDA_VISIBLE_DEVICES"] = "0"

claims = ["The students are preparing for an examination.", "The students are on vacation."]

doc = "A group of students gather in the school library to study for their upcoming final exams."

# model_name can be one of the followings:
# ['roberta-large', 'deberta-v3-large', 'flan-t5-large', 'Bespoke-MiniCheck-7B']

#  MiniCheck-Flan-T5-Large (770M) is the best fack-checking model 
# with size < 1B and reaches GPT-4 performance.
model_name = 'flan-t5-large'
print(f"Loading model {model_name}...")
scorer = MiniCheck(model_name=model_name, cache_dir='./ckpts')
pred_label, raw_prob, _, _ = scorer.score(docs=[doc, doc], claims=claims)
print(raw_prob)
f = lambda x:round(x*100,2)
prob_percentages = [f(x) for x in raw_prob]
print(f'Evaluating using model: {model_name}')
for idx, claim in enumerate(claims):
    print(f'Prediction for claim: \n"{claim}": {bool(pred_label[idx])}')
    print(f'Truthful probability: {prob_percentages[idx]}%')

print("Cleaning up Model 1...")
del scorer
del pred_label
del raw_prob

gc.collect()

torch.cuda.empty_cache()

model_name = 'Bespoke-MiniCheck-7B'
print(f"Memory cleared. Loading {model_name}...")

# Include bespoke model (Currently there are errors, use API instead?)
# # Alternatively, you can use our Bespoke-MiniCheck-7B model (7B) for evaluation. 
# # Bespoke-MiniCheck-7B is the most performant fact-checking model 
# # in the MiniCheck series AND is the current SOTA regardless of size.
# # It's also commercially useable! 
# # For commercial licensing, please contact company@bespokelabs.ai
scorer = MiniCheck(model_name=model_name, enable_prefix_caching=False, cache_dir='./ckpts')
pred_label, raw_prob, _, _ = scorer.score(docs=[doc, doc], claims=claims)

print(raw_prob)
f = lambda x:round(x*100,2)
prob_percentages = [f(x) for x in raw_prob]
print(f'\nEvaluating using model: {model_name}')
for idx, claim in enumerate(claims):
    print(f'Prediction for claim: \n"{claim}": {bool(pred_label[idx])}')
    print(f'Truthful probability: {prob_percentages[idx]}%')
```
